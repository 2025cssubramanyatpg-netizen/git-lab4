```python
import numpy as np
import pandas as pd

print("The version of numpy:", np.__version__)
print("The version of pandas:", pd.__version__)

marks = np.array([78, 85, 92, 67, 88])

print("Marks:", marks)
print("Type:", type(marks))
print("Size:", marks.size)

print("Original:", marks)
print("Add 5:", marks + 5)
print("Multiply by 2:", marks * 2)
print("Divide by 2:", marks / 2)
print("Square:", marks ** 2)

print("Sum:", marks.sum())
print("Mean:", marks.mean())
print("Maximum:", marks.max())
print("Minimum:", marks.min())

print("First mark:", marks[0])
print("Second mark:", marks[1])
print("Last mark:", marks[-1])

print("Index 1 to 3:", marks[1:4])
print("First three:", marks[:3])
print("From index 2:", marks[2:])
print("Every second element:", marks[::2])

student_marks = np.array([
    [78, 85, 92],
    [67, 88, 76],
    [90, 95, 89]
])

print(student_marks)
print("Shape:", student_marks.shape)

print("Mean of all values:", student_marks.mean())
print("Column-wise mean:", student_marks.mean(axis=0))
print("Row-wise mean:", student_marks.mean(axis=1))

data = {
    "Name": ["Annu", "Rahul", "Kiran", "Priya"],
    "Branch": ["CSE", "AIML", "CSE", "AIML"],
    "Marks": [85, 91, 76, 88]
}

df = pd.DataFrame(data)

print(df)
print("Shape:", df.shape)
print("Columns:", df.columns)

print(df.head())
print(df.tail())

df.info()
print(df.describe())

print("Marks Column:")
print(df["Marks"])

print("Branch and Marks:")
print(df[["Branch", "Marks"]])

print("First row:")
print(df.iloc[0])

print("Second row:")
print(df.iloc[1])

print("First three rows:")
print(df.iloc[:3])

print("Students scoring above 90:")
print(df[df["Marks"] > 90])

print("Students from CSE:")
print(df[df["Branch"] == "CSE"])

print("Students scoring 85 or above:")
print(df[df["Marks"] >= 85])

data = {
    "Name": ["Annu", "Bharat", "Chetan", "Divya", "Esha"],
    "Branch": ["CSE", "ISE", "CSE", "AIML", "ISE"],
    "Marks": [72, 91, 84, 95, 68],
    "Attendance": [82, 94, 88, 96, 79]
}

students = pd.DataFrame(data)

print(students)

print("Name and Marks:")
print(students[["Name", "Marks"]])

print("Students with marks more than 80:")
print(students[students["Marks"] > 80])

print("Students with attendance more than 85:")
print(students[students["Attendance"] >= 85])

print(students[0:3])

print("Average marks:")
print(students["Marks"].mean())

print("Maximum marks:")
print(students["Marks"].max())

print("Students from CSE:")
print(students[students["Branch"] == "CSE"])

print("Students with marks above 80 and attendance above 85:")
print(students[
    (students["Marks"] >= 80) &
    (students["Attendance"] >= 85)
])
```
